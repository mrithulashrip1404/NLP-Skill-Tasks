```python
import pandas as pd
import re
import string
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('wordnet', quiet=True)

df = pd.read_csv("IMDB Dataset.csv")

print("=" * 60)
print("CUSTOMER REVIEW CLEANING SYSTEM")
print("=" * 60)

print("\nDataset Loaded Successfully")
print("Shape:", df.shape)

print("\nFirst 5 Rows:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Reviews:")
print(df['review'].duplicated().sum())

def clean_text(text):
    text = text.lower()
    text = re.sub(r'<.*?>', ' ', text)
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)
    text = re.sub(r'\d+', ' ', text)
    text = text.translate(
        str.maketrans('', '', string.punctuation)
    )
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['cleaned_text'] = df['review'].apply(clean_text)

print("\nText Cleaning Completed")

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    words = [
        word for word in words
        if word not in stop_words
    ]
    return ' '.join(words)

df['without_stopwords'] = df['cleaned_text'].apply(
    remove_stopwords
)

print("Stopword Removal Completed")

df['tokens'] = df['without_stopwords'].apply(
    word_tokenize
)

print("Tokenization Completed")

lemmatizer = WordNetLemmatizer()

def lemmatize_tokens(tokens):
    return [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

df['lemmatized_tokens'] = df['tokens'].apply(
    lemmatize_tokens
)

print("Lemmatization Completed")

df['cleaned_review'] = df['lemmatized_tokens'].apply(
    lambda tokens: ' '.join(tokens)
)

print("Final cleaned_review column created")

print("\n" + "=" * 60)
print("ORIGINAL VS CLEANED REVIEWS")
print("=" * 60)

for i in range(3):
    print("\nOriginal Review:")
    print(df.loc[i, 'review'][:500])

    print("\nCleaned Review:")
    print(df.loc[i, 'cleaned_review'][:500])

    print("\n" + "-" * 60)

print("\nFinal Dataset:")
print(
    df[
        ['review', 'sentiment', 'cleaned_review']
    ].head()
)

df.to_csv(
    "cleaned_IMDB_reviews.csv",
    index=False
)

print("\n" + "=" * 60)
print("TASK COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nPreprocessed dataset saved as:")
print("cleaned_IMDB_reviews.csv")
```
