```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


print("=" * 60)
print("ASSIGNMENT SIMILARITY / PLAGIARISM DETECTOR")
print("=" * 60)


# Get user input
document1 = input("\nEnter Assignment 1: ")
document2 = input("Enter Assignment 2: ")


# Create DataFrame
data = pd.DataFrame({
    "Document": ["Assignment 1", "Assignment 2"],
    "Text": [document1, document2]
})


# Text preprocessing
data["Text"] = (
    data["Text"]
    .str.lower()
    .str.replace(r"[^a-z\s]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)


# TF-IDF
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(
    data["Text"]
)


# Cosine similarity
similarity_matrix = cosine_similarity(
    tfidf_matrix
)

similarity_score = similarity_matrix[0, 1]


# Percentage
similarity_percentage = similarity_score * 100


# Result
print("\n" + "=" * 60)
print("SIMILARITY RESULT")
print("=" * 60)

print(f"\nCosine Similarity : {similarity_score:.4f}")
print(f"Similarity Percentage : {similarity_percentage:.2f}%")


# Similarity level
if similarity_percentage >= 70:
    level = "High Similarity"
elif similarity_percentage >= 40:
    level = "Moderate Similarity"
else:
    level = "Low Similarity"


print(f"Similarity Level : {level}")

print("\n" + "=" * 60)
print("PLAGIARISM ANALYSIS COMPLETED")
print("=" * 60)
```
